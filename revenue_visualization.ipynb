```python
import pandas as pd

df = pd.read_csv('../data/category_growth.csv')
df.head()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))
plt.bar(df['product_category_name_english'], df['revenue_period1'], label='Period 1')
plt.bar(df['product_category_name_english'], df['revenue_period2'], label='Period 2', alpha=0.6)
plt.xticks(rotation=90)
plt.legend()
plt.title('Revenue Comparison by Category: Period 1 vs Period 2')
plt.xlabel('Product Category')
plt.ylabel('Revenue')
plt.show()
```

```python

df_sorted = df.sort_values('growth_pct')

colors = ['red' if x < 0 else 'green' for x in df_sorted['growth_pct']]

plt.figure(figsize=(12, 8))
plt.barh(df_sorted['product_category_name_english'], df_sorted['growth_pct'], color=colors)
plt.axvline(x=0, color='black', linewidth=0.8)
plt.title('Revenue Growth % by Category (Period 1 vs Period 2)')
plt.xlabel('Growth %')
plt.ylabel('Product Category')
plt.tight_layout()
plt.show()
```

```python

df_filtered = df_sorted[df_sorted['product_category_name_english'] != 'construction_tools_construction']

colors = ['red' if x < 0 else 'green' for x in df_filtered['growth_pct']]

plt.figure(figsize=(12, 8))
plt.barh(df_filtered['product_category_name_english'], df_filtered['growth_pct'], color=colors)
plt.axvline(x=0, color='black', linewidth=0.8)
plt.title('Revenue Growth % by Category (excluding construction_tools_construction outlier)')
plt.xlabel('Growth %')
plt.ylabel('Product Category')
plt.tight_layout()
plt.savefig('../reports/revenue_growth_chart.png', dpi=300, bbox_inches='tight')
plt.show()
```
